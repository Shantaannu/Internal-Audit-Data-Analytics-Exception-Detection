# Internal Audit Data Analytics & Exception Detection

## 2. Data Quality

### Objective

The objective of this section is to assess the completeness, consistency,
validity and integrity of the transaction data before performing internal
audit testing.

```python
import pandas as pd
import numpy as np
```

```python
transactions = pd.read_csv(
    "../data/raw/banking_transactions.csv",
    parse_dates=["Transaction_Date"]
)

employees = pd.read_csv(
    "../data/raw/employee_master.csv"
)

vendors = pd.read_csv(
    "../data/raw/vendor_master.csv"
)

thresholds = pd.read_csv(
    "../data/raw/approval_thresholds.csv"
)
```

```python
#Duplicate Transaction IDs
transactions["Transaction_ID"].duplicated().sum()
```

```python
transactions[
    transactions["Transaction_ID"].duplicated(keep=False)
].sort_values("Transaction_ID").head(20)
```

```python
#Exact Duplicate Records
transactions.duplicated().sum()
```

```python
transactions[
    transactions.duplicated(keep=False)
].head(20)
```

```python

#Missing Values
missing = transactions.isnull().sum()

missing[missing > 0]
```

```python
#Validate Categorical Values
transactions["Transaction_Type"].unique()
```

```python
transactions["Approval_Status"].unique()
```

```python
transactions["Channel"].unique()
```

```python
transactions["Branch"].unique()
```

```python
#Transaction Amount Validation
transactions["Amount_EUR"].describe()
```

```python
(transactions["Amount_EUR"] <= 0).sum()
```

```python
transactions[
    transactions["Amount_EUR"] <= 0
]
```

```python
transactions.nlargest(
    20, 
    "Amount_EUR"
)[
    ["Transaction_ID", "Transaction_Date", "Transaction_Type", "Amount_EUR"]
]
```

```python
#Date & Time Validation
transactions["Transaction_Date"].min()
```

```python
transactions["Transaction_Date"].max()
```

```python
transactions["Transaction_Date"].is_monotonic_increasing
```

```python
#Referential Integrity
invalid_employees = transactions[
    ~transactions["Employee_ID"].isin(employees["Employee_ID"])
]

invalid_employees.shape
```

```python
invalid_vendors = transactions[
    ~transactions["Vendor_ID"].isin(vendors["Vendor_ID"])
]

invalid_vendors.shape
```
